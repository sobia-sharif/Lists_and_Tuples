```python
# nested dictionary
student= {
    "name": "Sana",
    "marks": {
        "phy" : 45,
        "chem" : 30,
        "math" : 50,
    }

}
print(student)
print(student["marks"])      # nesting
print(student["marks"]["math"])

print(student.keys())     #method for getting keys only from data
print(list(student.keys()))   #if you want keys in list
print(len(list(student.keys())))  #give length of keys

#values methods

print(student.values())
print(len(student.values()))

#new dict
new_dict = {"education": "graduation", "gender": "female" } # new information
student.update(new_dict)   # use method to incorporate new dict
print(student)    # then print
```
